if-elif-else

```python
# Exam results
# A: 90
# B: 80
# C: 70
# D: 60
# E: 50
# F: <50
#
# let user input a score

score = 60
if score >= 90:
        print("A")
elif score >= 80:
        print("B")
elif score >= 70:
        print("C")
elif score >= 60:
        print("D")
elif score >= 50:
        print("E")
else:
        print ("F")
```

```python
# 9. Köpcheck

pengar_pa_kontot = 260
pris_pa_varan = 20

# Kolla om du har råd
if pengar_pa_kontot >= pris_pa_varan:
    # Räkna ut hur mycket som blir kvar
    kvarvarande_pengar = pengar_pa_kontot - pris_pa_varan
    print("Du har råd! Du har " + str(kvarvarande_pengar) + " kr kvar på kontot.")
else:
    print("Tyvärr, du har inte råd med varan.")
```

```python
# 11. Slututmaning: Biobiljetten

# Steg 1: Fråga efter ålder
alder = int(input("Vänligen skriv in din ålder: "))

# Steg 2: Fråga om medlemskap
medlem = input('Är du medlem i bioklubben? (ja/nej): ')

# Steg 3: Kolla om något av villkoren för rabatt är uppfyllt
if alder < 15 or alder >= 65 or medlem == "ja":
    # Steg 4: Om villkoret stämmer
    print("Du får rabatt på biljetten")
else:
    # Om inget av villkoren stämmer
    print("Fullt pris, 120 kr.")
```

10 enkla övningar i for-satser

```python
# 1. Skriv ut talen 1 till 5:

tal = [1, 2, 3, 4, 5]

for talen in tal:
    print(tal)
```

```python
# 2. Skriv ut en lista med frukter:
frukter = ["äpple", "banan", "apelsin"]

for frukt in frukter:
    print(frukt)
```

```python
# 3. Summera tal:

# 1. Skapa listan med tal
tal_lista = [2, 4, 6, 8]

# 2. Skapa en variabel som börjar på 0 (vår "tomma plånbok")
summa = 0

# 3. Starta loopen
for tal in tal_lista:
    # För varje varv lägger vi till det nuvarande talet till summan
    summa = summa + tal

# 4. Skriv ut slutresultatet efter loopen
print("Den totala summan är:", summa)
```
